```python

import pandas as pd
import numpy as np
```

```python

url = "https://raw.githubusercontent.com/shaadyalii/Dirty-Cafe-Sales-Dataset/main/dirty_cafe_sales.csv"

df = pd.read_csv(url)

df.head()
```

```python
# Obtener el número de columnas
num_columnas = df.shape[1]
print(f"La base de datos tiene {num_columnas} columnas.")

# También puedes listar los nombres de las columnas para verificarlas
print("Columnas:", list(df.columns))

# Muestra tipos de datos
tipo_de_datos = df.dtypes

print((tipo_de_datos))


#Muestra información general del dataframe
df.info
```

```python
# Eliminar valores nulos
df.isnull ().sum()

```

```python
#Identificar duplicados
df.duplicated() . sum()
```

```python
#En qué columnas aparencen y cuantos hay
for col in df.columns:
  error = df[col].isin(["ERROR", "UNKNOWN"]).sum()
  if error > 0:
    print(col, error)
```

```python
#Reemplaza error
df = df.replace(["ERROR", "UNKNOWN"], np.nan)
#Revisar valores nulos otra vez
df.isnull().sum()
```

```python
#Eliminar duplicados
filas_originales= len(df)
df= df.drop_duplicates()
filas_despues =len (df)
print("Filas originales", filas_originales)
print (" Filas despues", filas_despues)
print("Filas duplicadas", filas_originales -  filas_despues)

```

```python
#Convertir datos enteros
columnas_numericas = ["Quantity", "Price Per Unit", "Total Spent"]
for col in columnas_numericas:
  df[col] = pd.to_numeric(df[col], errors="coerce")

df[columnas_numericas].dtypes
```

```python
#Convertir formatos de fecha
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"], errors = "coerce")
df["Transaction Date"].dtypes
```

```python
#Verificar información de nuevo
df.info()
```

```python
#Tratamiento de datos faltantes
# Completar el texto faltante en Item
df["Item"] = df["Item"].fillna("Unknown")
print("Nulos en Item:", df["Item"].isna().sum())

# Forzar la conversión de Quantity a numérico por seguridad
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

# Calcular la mediana de Quantity
mediana_cantidad = df["Quantity"].median()

# Completar valores faltantes de Quantity usando la mediana calculada
df["Quantity"] = df["Quantity"].fillna(mediana_cantidad)
print("Nulos en Quantity:", df["Quantity"].isna().sum())
```

```python
# Reemplazar valores de error por NaN para limpiar los datos
df = df.replace(["ERROR", "UNKNOWN"], np.nan)

# Asegurar la conversión a tipo numérico de 'Price Per Unit'
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"], errors="coerce")

# Calcular la mediana ignorando valores nulos usando NumPy
mediana_precio_np = np.nanmedian(df["Price Per Unit"])

# Reemplazar los valores faltantes con la mediana obtenida
df["Price Per Unit"] = df["Price Per Unit"].fillna(mediana_precio_np)

print("Mediana calculada con NumPy:", mediana_precio_np)
print("Nulos restantes en Price Per Unit:", df["Price Per Unit"].isna().sum())
```

```python
# Asegurar que 'Quantity' y 'Price Per Unit' estén en formato numérico
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"], errors="coerce")

# Reconstruir 'Total Spent' calculando el producto de Quantity por Price Per Unit
df["Total Spent"] = df["Quantity"] * df["Price Per Unit"]

# Mostrar estadísticas de la reconstrucción
print("Nulos restantes en Total Spent:", df["Total Spent"].isna().sum())
display(df[["Quantity", "Price Per Unit", "Total Spent"]].head())
```

```python
# Columnas categóricas a tratar
columnas_categoricas = ["Payment Method", "Location"]

for col in columnas_categoricas:
    # Calcular la moda (usamos [0] ya que .mode() devuelve una Serie)
    moda = df[col].mode()[0]

    # Rellenar los valores nulos con la moda
    df[col] = df[col].fillna(moda)

    print(f"La moda para '{col}' es: {moda}")
    print(f"Nulos restantes en '{col}': {df[col].isna().sum()}\n")
```

```python
# 1. Asegurar que la columna sea de tipo fecha (datetime)
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"], errors="coerce")

# 2. Eliminar temporalmente los NaT y ordenar las fechas válidas
fechas_validas = df["Transaction Date"].dropna().sort_values()

# 3. Obtener el valor central (la mediana de las fechas)
indice_central = len(fechas_validas) // 2
fecha_central = fechas_validas.iloc[indice_central]

# 4. Usar fillna() para completar los valores faltantes
df["Transaction Date"] = df["Transaction Date"].fillna(fecha_central)

print(f"La fecha central representativa es: {fecha_central.date()}")
print("Nulos restantes en Transaction Date:", df["Transaction Date"].isna().sum())
```

```python
# Calcular la nueva variable 'Revenue_Calculated'
df["Revenue_Calculated"] = df["Quantity"] * df["Price Per Unit"]

# Mostrar las primeras filas para verificar el resultado
display(df[["Quantity", "Price Per Unit", "Revenue_Calculated"]].head())
print("Valores nulos en Revenue_Calculated:", df["Revenue_Calculated"].isna().sum())
```

```python
# Comprobar si Revenue_Calculated contiene valores nulos
nulos_revenue = df["Revenue_Calculated"].isna().sum()
print(f"El número de valores nulos en 'Revenue_Calculated' es: {nulos_revenue}")
```

```python
# Crear la columna 'Revenue_Level' usando np.where()
df["Revenue_Level"] = np.where(df["Revenue_Calculated"] >= 15, "High", "Low")

# Mostrar una muestra de los datos para observar la clasificación
display(df[["Revenue_Calculated", "Revenue_Level"]].head(10))
```

```python
# Verificar cuántos valores nulos quedan en cada columna del DataFrame
valores_nulos_final = df.isnull().sum()
print(valores_nulos_final)
```

```python
# 1. Agrupar por producto ('Item') y sumar la cantidad vendida ('Quantity')
producto_mas_vendido = df.groupby("Item")["Quantity"].sum().idxmax()
cantidad_maxima = df.groupby("Item")["Quantity"].sum().max()

# 2. Agrupar por producto ('Item') y sumar el revenue calculado ('Revenue_Calculated')
producto_mas_revenue = df.groupby("Item")["Revenue_Calculated"].sum().idxmax()
revenue_maximo = df.groupby("Item")["Revenue_Calculated"].sum().max()

print(f"1. El producto con mayor cantidad vendida es: '{producto_mas_vendido}' con un total de {cantidad_maxima:.0f} unidades.")
print(f"2. El producto que genera mayor revenue es: '{producto_mas_revenue}' con un total de ${revenue_maximo:,.2f}.")
```
