```python
# Limpieza de datos
#Reemplazo de etiquetas invalidas


df_animales = df_animales.replace(
    ["ERROR", "UNKNOWN", "not determined"],
    np.nan
)

print("--- VALORES NULOS DESPUÉS DEL REEMPLAZO ---")
print(df_animales.isnull().sum())
```

```python

# Convertir los valores a minúsculas y eliminar espacios
df_animales["Animal type"] = (
    df_animales["Animal type"]
    .str.strip()
    .str.lower()
)

# Diccionario de correcciones
correcciones_animales = {
    "european bison™": "european bison",
    "european bisson": "european bison",
    "european buster": "european bison",
    "red squirrell": "red squirrel",
    "red squirel": "red squirrel",
    "lynx?": "lynx",
    "ledgehod": "hedgehog",
    "wedgehod": "hedgehog"
}

df_animales["Animal type"] = (
    df_animales["Animal type"]
    .replace(correcciones_animales)
)

print("--- ANIMAL TYPE ESTANDARIZADO ---")
print(df_animales["Animal type"].value_counts(dropna=False))
```

```python
correcciones_paises = {
    "PL": "Poland",
    "HU": "Hungary",
    "Hungry": "Hungary",
    "CZ": "Czech Republic",
    "Czech": "Czech Republic",
    "DE": "Germany"
}

df_animales["Country"] = (
    df_animales["Country"]
    .str.strip()
    .replace(correcciones_paises)
)

# Valores sospechosos o no verificables
# Se clasifican como desconocidos para evitar
# asignar un país incorrecto.

df_animales["Country"] = df_animales["Country"].replace(
    ["CC", "Australia"],
    "Unknown Country"
)

print("\n--- COUNTRY ESTANDARIZADO ---")
print(df_animales["Country"].value_counts(dropna=False))
```

Nota: Se clasificado CC y Australia como desconocidos porque sus registros presentan una inconsistencia con el contexto geográfico de las observaciones. No es correcto asignarles otro país sin evidencia suficiente. Convendría verificar esos registros con la fuente original.

```python
df_animales["Gender"] = (
    df_animales["Gender"]
    .str.strip()
    .str.lower()
)

print("\n--- CATEGORÍAS DE GENDER ---")
print(df_animales["Gender"].value_counts(dropna=False))
```

```python
#  12. CONVERSIÓN DE VARIABLES NUMÉRICAS


columnas_numericas = [
    "Weight kg",
    "Body Length cm",
    "Latitude",
    "Longitude"
]

for col in columnas_numericas:

    df_animales[col] = pd.to_numeric(
        df_animales[col],
        errors="coerce"
    )

print("--- TIPOS DE DATOS NUMÉRICOS ---")
print(df_animales[columnas_numericas].dtypes)


# 13. CONVERSIÓN DE FECHAS

df_animales["Observation date"] = pd.to_datetime(
    df_animales["Observation date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("\nTipo de dato de Observation date:")
print(df_animales["Observation date"].dtype)
```

```python
# 14. IDENTIFICAR Y CORREGIR MEDIDAS NEGATIVAS

# Los pesos y longitudes negativos se consideran inválidos.

df_animales.loc[
    df_animales["Weight kg"] < 0,
    "Weight kg"
] = np.nan

df_animales.loc[
    df_animales["Body Length cm"] < 0,
    "Body Length cm"
] = np.nan

print("--- VERIFICACIÓN DE VALORES NEGATIVOS ---")

print(
    "Pesos negativos restantes:",
    (df_animales["Weight kg"] < 0).sum()
)

print(
    "Longitudes negativas restantes:",
    (df_animales["Body Length cm"] < 0).sum()
)
```

```python
#Emilinar columnas vacias

print("Columnas originales:", df_animales.shape[1])

df_animales = df_animales.drop(
    columns=["Animal code"]
)

print("Columnas después de eliminar:", df_animales.shape[1])
```

```python
# Eliminar duplicados

filas_originales = len(df_animales)

df_animales = df_animales.drop_duplicates()

filas_despues = len(df_animales)

print("--- RESULTADO DE ELIMINACIÓN DE DUPLICADOS ---")

print("Filas originales:", filas_originales)

print("Filas después de limpiar:", filas_despues)

print("Duplicados eliminados:",
      filas_originales - filas_despues)

print("Duplicados restantes:",
      df_animales.duplicated().sum())
```

```python
#Imputacion de variables categoricas
# Animal type
df_animales["Animal type"] = df_animales[
    "Animal type"
].fillna("Unknown Animal")

# Country
df_animales["Country"] = df_animales[
    "Country"
].fillna("Unknown Country")

# Gender
df_animales["Gender"] = df_animales[
    "Gender"
].fillna("Unknown")

print("--- VERIFICACIÓN DE VARIABLES CATEGÓRICAS ---")

for col in ["Animal type", "Country", "Gender"]:

    print(
        f"Nulos restantes en {col}:",
        df_animales[col].isnull().sum()
    )


# 18. IMPUTACIÓN DE PESO Y LONGITUD
# UTILIZANDO LA MEDIANA POR TIPO DE ANIMAL


columnas_medidas = [
    "Weight kg",
    "Body Length cm"
]

for col in columnas_medidas:

    # Calcular mediana según el tipo de animal
    mediana_grupo = df_animales.groupby(
        "Animal type"
    )[col].transform("median")

    # Reemplazar nulos con la mediana de cada grupo
    df_animales[col] = df_animales[col].fillna(
        mediana_grupo
    )

    # Si todavía existen nulos, utilizar la mediana global
    mediana_global = df_animales[col].median()

    df_animales[col] = df_animales[col].fillna(
        mediana_global
    )

print("\n--- MEDIDAS FÍSICAS DESPUÉS DE IMPUTAR ---")

print(df_animales[columnas_medidas].isnull().sum())


#Verificar variables

display(
    df_animales[
        ["Animal type", "Weight kg", "Body Length cm"]
    ].head(10)
)
```

```python
# Validacion final

print("--- DIMENSIONES FINALES ---")
print(df_animales.shape)

print("\n--- TIPOS DE DATOS FINALES ---")
print(df_animales.dtypes)

print("\n--- VALORES NULOS FINALES ---")
print(df_animales.isnull().sum())

print("\n--- DUPLICADOS RESTANTES ---")
print(df_animales.duplicated().sum())

print("\n--- CATEGORÍAS ESTANDARIZADAS ---")

for col in ["Animal type", "Country", "Gender"]:

    print(f"\n{col}:")
    print(df_animales[col].value_counts(dropna=False))

print("\n--- INFORMACIÓN GENERAL FINAL ---")
df_animales.info()
```

```python
#CLASIFICACIÓN DE ANIMALES SEGÚN SU PESO

# Calcular la mediana del peso
mediana_peso = np.nanmedian(df_animales["Weight kg"])

print("Mediana del peso:", mediana_peso)

# Crear nueva variable utilizando np.where()
df_animales["Weight Category"] = np.where(
    df_animales["Weight kg"] >= mediana_peso,
    "Above or Equal to Median",
    "Below Median"
)

# Mostrar resultados
display(
    df_animales[
        ["Animal type", "Weight kg", "Weight Category"]
    ].head(10)
)

# Contar las categorías
print("\n--- DISTRIBUCIÓN DE CATEGORÍAS DE PESO ---")

print(df_animales["Weight Category"].value_counts())
```

```python
# Media del peso
media_peso = np.nanmean(df_animales["Weight kg"])

# Mediana del peso
mediana_peso = np.nanmedian(df_animales["Weight kg"])

# Desviación estándar del peso
desviacion_peso = np.nanstd(df_animales["Weight kg"])

print("--- ESTADÍSTICAS DEL PESO DE LOS ANIMALES ---")

print(f"Media del peso: {media_peso:.2f} kg")

print(f"Mediana del peso: {mediana_peso:.2f} kg")

print(f"Desviación estándar: {desviacion_peso:.2f} kg")
```

```python
# CANTIDAD DE OBSERVACIONES POR ANIMAL


observaciones_animal = (
    df_animales.groupby("Animal type")
    .size()
    .sort_values(ascending=False)
)

print("--- OBSERVACIONES POR TIPO DE ANIMAL ---")

print(observaciones_animal)
```

```python
# Peso promedio por animal

peso_promedio_animal = (
    df_animales.groupby("Animal type")["Weight kg"]
    .mean()
    .sort_values(ascending=False)
)

print("--- PESO PROMEDIO POR TIPO DE ANIMAL ---")

print(peso_promedio_animal.round(2))
```

```python
# Peso promedio segun pais
peso_promedio_pais = (
    df_animales.groupby("Country")["Weight kg"]
    .mean()
    .sort_values(ascending=False)
)

print("--- PESO PROMEDIO POR PAÍS ---")

print(peso_promedio_pais.round(2))
```

```python
# Longitud promedio por tipo de animal

longitud_promedio_animal = (
    df_animales.groupby("Animal type")["Body Length cm"]
    .mean()
    .sort_values(ascending=False)
)

print("--- LONGITUD PROMEDIO POR TIPO DE ANIMAL ---")

print(longitud_promedio_animal.round(2))
```

```python
# DISTRIBUCIÓN POR GÉNERO Y TIPO DE ANIMAL


distribucion_genero = pd.crosstab(
    df_animales["Animal type"],
    df_animales["Gender"]
)

print("--- DISTRIBUCIÓN POR GÉNERO Y ESPECIE ---")

display(distribucion_genero)
```

```python

# 1. Especie con mayor número de observaciones

especie_mas_observada = observaciones_animal.idxmax()

cantidad_observaciones = observaciones_animal.max()


# 2. Especie con mayor peso promedio

especie_mayor_peso = peso_promedio_animal.idxmax()

peso_maximo_promedio = peso_promedio_animal.max()


# 3. Especie con mayor longitud promedio

especie_mayor_longitud = longitud_promedio_animal.idxmax()

longitud_maxima_promedio = longitud_promedio_animal.max()


# 4. País con mayor peso promedio
# Excluimos Unknown Country porque no representa
# una ubicación geográfica identificada.

peso_paises_validos = peso_promedio_pais[
    peso_promedio_pais.index != "Unknown Country"
]

pais_mayor_peso = peso_paises_validos.idxmax()

peso_promedio_mayor_pais = peso_paises_validos.max()


# ==========================================
# MOSTRAR CONCLUSIONES
# ==========================================

print("========== CONCLUSIONES DEL ANÁLISIS ==========\n")

print(
    f"1. La especie con mayor número de observaciones es "
    f"{especie_mas_observada}, con {cantidad_observaciones} registros. "
    f"Esto indica que es la especie con mayor representación "
    f"dentro del dataset."
)

print(
    f"\n2. La especie con mayor peso promedio es "
    f"{especie_mayor_peso}, con {peso_maximo_promedio:.2f} kg. "
    f"Esto evidencia diferencias importantes en las dimensiones "
    f"físicas entre las especies analizadas."
)

print(
    f"\n3. La especie con mayor longitud corporal promedio es "
    f"{especie_mayor_longitud}, con {longitud_maxima_promedio:.2f} cm. "
    f"Esto permite identificar qué especie presenta las mayores "
    f"dimensiones longitudinales en promedio."
)

print(
    f"\n4. El país con mayor peso promedio de los animales registrados "
    f"es {pais_mayor_peso}, con {peso_promedio_mayor_pais:.2f} kg. "
    f"Este resultado debe interpretarse considerando la composición "
    f"de especies y el número de observaciones de cada país."
)
```
